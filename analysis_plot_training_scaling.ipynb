```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# 模拟数据
K = [6, 8, 10, 12]
# datasets = ["TriviaQA", "SciQ", "NQ-Open", "WebQuestions"]
datasets = ["TriviaQA", "SciQ", "NQ-Open", "WebQ-QA"]

# Precision数据: k=6,8,10,12
precision_data = {
    "TriviaQA":     [82.95, 84.36, 84.01, 84.36],
    "SciQ":         [80.29, 81.52, 80.94, 80.76],
    "NQ-Open":      [57.99, 60.95, 60.44, 60.35],
    "WebQ-QA":      [67.31, 67.42, 67.1, 66.15],
}

# Truthfulness数据: k=6,8,10,12
truthfulness_data = {
    "TriviaQA":     [59.8, 63.55, 63.79, 64.65],
    "SciQ":         [49.7, 54.70, 55.2, 55.4],
    "NQ-Open":      [26.52, 25.51, 25.98, 26.4],
    "WebQ-QA":      [33.75, 37.76, 38.43, 38.28],
}

# 颜色 & marker
# colors = sns.color_palette("Set1", n_colors=4)
markers = ["o", "s", "D", "^"]

# 创建两个子图 (并排 1×2)
c = 0.5
fig, axs = plt.subplots(1, 2, figsize=(12*c, 5*c), sharey=True)

# Precision 折线图
for i, ds in enumerate(datasets):
    axs[0].plot(K, 
                precision_data[ds], 
                marker=markers[i], 
                # color=colors[i], 
                color=f"C{i}",   # 使用 matplotlib 默认颜色 
                label=ds, 
                linewidth=1.5,   # 默认是2
                markersize=5      # 默认是6
                )
# axs[0].set_xlabel("K", fontsize=10)
axs[0].set_ylabel("Precision (%)", fontsize=10, labelpad=-5) # 用labelpad调整标签与轴的距离,默认值是4～5
axs[0].set_ylim(0, 100)   # y轴范围 0-100
axs[0].legend(loc="lower right", fontsize=9, ncol=2,
              columnspacing=0.8,   # 列间距（默认 2.0）
              handletextpad=0.4,   # 图例标记和文字的间距（默认 0.8）
              borderaxespad=0.3    # 图例和坐标轴之间的间距（默认 0.5）
              )
axs[0].grid(True, linestyle="--", alpha=0.6)
# axs[0].set_title("Precision vs K")

# Truthfulness 折线图
for i, ds in enumerate(datasets):
    axs[1].plot(K, 
                truthfulness_data[ds], 
                marker=markers[i], 
                # color=colors[i], 
                color=f"C{i}",   # 使用 matplotlib 默认颜色 
                label=ds, 
                linewidth=1.5,   # 默认是2
                markersize=5      # 默认是6
                )
# axs[1].set_xlabel("K", fontsize=10)
axs[1].set_ylabel("Truthfulness (%)", fontsize=10)
axs[1].set_ylim(0, 100)   # y轴范围 0-100
axs[1].legend(loc="lower right", fontsize=9, ncol=2,
              columnspacing=0.8,   # 列间距（默认 2.0）
              handletextpad=0.4,   # 图例标记和文字的间距（默认 0.8）
              borderaxespad=0.3    # 图例和坐标轴之间的间距（默认 0.5）
             )
axs[1].grid(True, linestyle="--", alpha=0.6)
# axs[1].set_title("Truthfulness vs K")

# 调整布局
plt.tight_layout()
# plt.tight_layout(rect=[0, 0.1, 1, 1])
# plt.show()
plt.savefig('analysis_training_scaling_xn_width15.pdf', dpi=300)
# plt.savefig('analysis_training_scaling_xn.pdf', dpi=300)
```
